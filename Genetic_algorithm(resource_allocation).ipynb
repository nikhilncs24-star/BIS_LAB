```python
import random

# Project attributes: (Profit, Budget, Manpower)
PROJECTS = [(80, 20, 4), (40, 15, 2), (100, 35, 7), (60, 10, 3), (120, 40, 8), (90, 25, 5), (50, 12, 2)]
MAX_BUDGET, MAX_MANPOWER = 70, 15

def get_fitness(chrom):
    """Calculates profit; returns 0 if constraints are violated or nothing is selected."""
    if not any(chrom): return 0
    p, b, m = zip(*[PROJECTS[i] for i, gene in enumerate(chrom) if gene])
    return sum(p) if (sum(b) <= MAX_BUDGET and sum(m) <= MAX_MANPOWER) else 0

def run_ga(pop_size=30, generations=40, mut_rate=0.1):
    pop = [[random.randint(0, 1) for _ in PROJECTS] for _ in range(pop_size)]
    
    for _ in range(generations):
        select = lambda: max(random.choices(pop, k=3), key=get_fitness)
        next_pop = []
        while len(next_pop) < pop_size:
            p1, p2 = select(), select()
            pt = random.randint(1, len(PROJECTS) - 1)
            c1, c2 = p1[:pt] + p2[pt:], p2[:pt] + p1[pt:]
            next_pop.extend([[1 - g if random.random() < mut_rate else g for g in c] for c in (c1, c2)])
            
        pop = next_pop[:pop_size]
        
    best = max(pop, key=get_fitness)
    return best, get_fitness(best)

# Run and print results
best_alloc, max_profit = run_ga()
print(f"Max Profit: {max_profit}")
print(f"Optimal Allocation Vector (0=No, 1=Yes): {best_alloc}")
```
